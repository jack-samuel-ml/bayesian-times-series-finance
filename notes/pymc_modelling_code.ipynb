### Below is how to implement a model where each parameter is its own distribution.

```python
import pymc as pm
import arviz as az
```

```python
data = [1.0, 2.0, 3.0]  # Example observed data
some_function_of = lambda param, data: param + sum(data)  # Example deterministic function

"""
IGNORE ABOVE

    This is a simple example of a probabilistic model using PyMC. The model consists of:
    1. A prior distribution for a parameter (param).
    2. A deterministic function that relates the parameter to the observed data.
    3. A likelihood that describes how the observed data is generated from the deterministic function.
"""

with pm.Model() as model:
    # 1. Define priors: parameters as distributions
    param = pm.Normal("param", mu=0, sigma=1)

    # 2. Deterministic relationship: here the (independent) parameter is related to the observed data through a function
    # mu is the output of this relationship (like many predicted means)
    mu = some_function_of(param, data)

    # 3. Likelihood: how probable the observed data is given the model parameters and deterministic function
    # this compares the observed (dependent values) data to the predicted means (mu) with some assumed noise (sigma)
    obs = pm.Normal("obs", mu=mu, sigma=1, observed=data)

    # 4. Sample the posterior
    # this iterates steps 1-3 many times to generate a distribution of likely parameter values given the observed data
    # a posterior distribution is generated for the parameter(s) of interest that is consistent with the observed data
    idata = pm.sample()
```
