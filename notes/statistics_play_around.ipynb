Below is an experimentation of the error function and its inverse.

```python
from scipy.special import erf, erfinv
import numpy as np
```

```python
erf_values = []
erfi_values = []

fraction_values = [0.9, 0.95, 0.99]

for i in range(1,4):
    erf_values.append(erf(i / np.sqrt(2)))

for frac in fraction_values:
    erfi_values.append(np.sqrt(2) *erfinv(frac))

erf_values_4sf = [f"{value:.4g}" for value in erf_values]
erfinv_values_4sf = [f"{value:.4g}" for value in erfi_values]

print(f"Erf values for 1, 2, and 3: {erf_values_4sf}")
print(f"Erfinv values for fractions 0.9, 0.95, and 0.99: {erfinv_values_4sf}")
```
